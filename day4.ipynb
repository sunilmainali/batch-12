```python
# series-one d data structures in pandas is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. A Series is like a column in a table. You can create a Series by passing a list or array of values to the pandas.Series() constructor.
# dataframe-two d data structures in pandas is a two-dimensional labeled data structure with columns of potentially different types. You can think of it like a spreadsheet or SQL table, or a dictionary of Series objects. You can create a DataFrame by passing a dictionary of lists or arrays to the pandas.DataFrame() constructor.
```

### indexing
 -accessing individual elements
 we are performing indexing using 
 loc and iloc 

### indexing using loc 

## loc in indexing  meaning location........we can index elements using label based indexing 
obj.loc[row level,col_level]

```python
df1
```

```python
df1.loc[1,'col1']
```

```python
df1.loc[9, 'col5']
 
```

```python
df1.loc[-1,'col5'] # we cannnot access using negative indexing in loc
```

## using iloc 
iloc-integer based location 

- obj.iloc[row_index,col_index]

```python
df1
```

```python
df1.iloc[2,1]
```

```python
df1.iloc[-1,-1]
```

### slicing in pandas 
- using loc 
- syntax obj.loc[row level slicing,col_level_slicing]

```python
df1
```

```python
df1.loc[:1,'col1':'col2']
 
```

```python
df.iloc[0:2, 0:2] 
```

```python
df1.loc[0:1, :'col2']
```
