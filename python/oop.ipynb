#### class
A class is a collection of objects. Classes are blueprints for creating objects. A class defines a set of attributes and methods that the created objects (instances) can have.

- Classes are created by keyword class.
- Attributes are the variables that belong to a class.
- Attributes are always public and can be accessed using the dot (.) operator. Example: Myclass.Myattribute


```python
### creating a class
class dog:
    species='german shepherd' #class attribute
    def __init__(self,name,age):
        self.name=name #instance attribute
        self.age=age #instance attribute
```

##### Explanation for above:
- class Dog: creates a class named Dog, which acts as a blueprint for dog objects.
- species is a class attribute, meaning it is shared by all instances of the class.
- __init__() is a constructor method that runs automatically when a new object is created. It is used to initialize object data.
- self refers to the current object, allowing each object to store and access its own data.
- self.name and self.age are instance attributes, unique to each Dog object created from the class.

##### Objects
An Object is an instance of a Class. It represents a specific implementation of the class and holds its own data. An object consists of:

- State: It is represented by the attributes and reflects the properties of an object.
- Behavior: It is represented by the methods of an object and reflects the response of an object to other objects.
- Identity: It gives a unique name to an object and enables one object to interact with other objects.

Creating an object involves instantiating a class to create a new instance of that class. This process is also referred to as object instantiation.

```python
### creating a class
class dog:
    species='german shepherd' #class attribute  ## Attributes → “What the object HAS” (data), 
    def __init__(self,name,age): # Methods → “What the object DOES” (functions),behaviour
        self.name=name #instance attribute
        self.age=age #instance attribute

# creating an object of the class
dog1=dog('tommy',3)
print(dog1.name)
print(dog1.age)
print(dog1.species)
```

```python
## create a class that can perform addition and subtraction of two numbers
class calculator:
    def __init__(self,num1,num2):
        self.num1=num1
        self.num2=num2
    def add(self):
        return self.num1+self.num2
    def subtract(self):
        return self.num1-self.num2
    
calc=calculator(10,5)
print(calc.add())
print(calc.subtract())

print(calculator.add(calc))
```

##### Types of attributes:
**1. Instance attributes**
- These are the attributes that belong to instances of a class
- They are defined within the constructor method **__init__** and can be accessed using the self keyword like self.name and self.age in the above Dog class.. 
- They are initialized when a new instance of the class is created.

**2. Class attributes**

- Class attributes are attributes that belong to the class itself. 
- They are defined outside the constructor method __init__ and can be accessed using the class name. 
- Class attributes are shared by all instances of the class 

##### Types of Methods

**1. Instance methods**
- The most common type of method in Python. These are the methods that operate on an instance of a class and have access to the instance's attributes.

**2. Class methods**
- Class methods are methods that operate on the class itself rather than on instances of the class
- They are defined using the @classmethod

**3. Static methods**
- Static methods are methods that do not operate on the instance or the class, but are related to the class in some way.
- They are defined using the @staticmethod decorator


```python
### creating a class
class dog:
    species='german shepherd' #class attribute
    def __init__(self,name,age):
        self.name=name #instance attribute
        self.age=age #instance attribute

    # instance method
    def bark(self):
        print(self.name+' is barking')
    
    #class method
    @classmethod
    def change_species(cls,new_species):
        cls.species=new_species

    # static method
    @staticmethod
    def info():
        print('Dogs are loyal animals')

    

# creating an object of the class
dog1=dog('tommy',3)
print(dog1.name) # instance attribute
print(dog1.age) # instance attribute
print(dog1.species)# class attribute

dog1.bark() # instance method
dog.change_species('bulldog') # class method
print(dog.species) # class attribute

dog.info() # static method
```

### four pillars of OOP

#### 1. Encapsulation (The "Black Box")
- **The Concept** :Keeping data and the methods that use that data grouped together in one unit, while hiding the internal complexity.(data hiding /protection)
- **The Example:** :When you drive a car, you use the steering wheel and pedals. You don’t need to know how the fuel injection system or the pistons are moving under the hood. The "guts" of the car are hidden (encapsulated) for safety and ease of use.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private attribute

    def show_balance(self):
        print("Current balance:", self.__balance)


acc = BankAccount(1000)
acc.show_balance()
print(acc.__balance)  # This will raise an AttributeError
```

#### 2. Abstraction (The "User Interface")
- **The Concept:** Abstraction hides the internal implementation details while exposing only the necessary functionality. It helps focus on "what to do" rather than "how to do it."
- **The Example:** Every car has a "Start" button or a key ignition. You know that pressing it starts the engine. You don't need a manual on internal combustion every time you want to go to the grocery store. Abstraction is the interface you interact with.

#### 3. Inheritance (The "Family Tree")
- **The Concept:** Creating a new class that "inherits" the features of an existing one.

- **The Example:** Think of a generic Vehicle. It has wheels and an engine. A Truck, a Sedan, and a Motorcycle are all "children" of the Vehicle. They inherit the wheels and engine, but add their own specific features (like a truck's flatbed or a motorcycle's kickstand).

```python
class Animal:

    def bark(self):
        print('Animal is barking')

class Dog(Animal):
    def eat(self):
        print('Dog is eating')

dog1=Dog()
dog1.eat()
dog1.bark()
```

#### 4. Polymorphism (The "Many Shapes")
- **The Concept:** The ability of different objects to respond to the same command in their own way.

- **The Example:** If you tell a Car to "Move," it rotates four wheels. If you tell a Boat to "Move," it spins a propeller. If you tell an Airplane to "Move," it engages jet engines. The command is the same ("Move"), but the implementation depends on the object.{same operation different behaviour}

#### INHERITANCE
Inheritance is a fundamental concept in object-oriented programming (OOP) that allows a class (called a child or derived class) to inherit attributes and methods from another class (called a parent or base class).

Here, a parent class Animal is created that has a method info(). Then a child classes Dog is created that inherit from Animal and add their own behavior.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal name:", self.name)

class Dog(Animal):
    def sound(self):
        print(self.name, "barks")

d = Dog("Buddy")
# Inherited method
d.info()     
d.sound()
```

Explanation:

- class Animal defines the parent class.
- info() prints the name of the animal.
- class Dog(Animal) defines Dog as a child of Animal class.
- d.info() calls parent method info() and d.sound() calls child method

Why do we need Inheritance
- Promotes code reusability by sharing attributes and methods across classes.
- Models real-world hierarchies like Animal -> Dog or Person -> Employee.
- Simplifies maintenance through centralized updates in parent classes.
- Enables method overriding for customized subclass behavior.
- Supports scalable, extensible design using polymorphism

Single Inheritance
- Single inheritance enables a derived class to inherit properties from a single parent class, thus enabling code reusability and the addition of new features to existing code.

```python
class Parent:
    def function1(self):
        print("This is function 1 from Parent class")

class child(Parent):
    def function2(self):
        print("This is function 2 from Child class")

child1=child()
child1.function1() # inherited method
child1.function2() # child class method
```

Multiple Inheritance
- When a class can be derived from more than one base class this type of inheritance is called multiple inheritances. 
- In multiple inheritances, all the features of the base classes are inherited into the derived class.

```python
class Mother:
    def func1(self):
        print("she is my mother")

class Father:
    def func2(self):
        print("he is my father")

class Child(Mother, Father):
    def func3(self):
        print("I am their child")

# create an object of Child class
obj = Child()

# access functions using the object
obj.func1()
obj.func2()
obj.func3()
```

Multilevel Inheritance
- In multilevel inheritance, features of the base class and the derived class are further inherited into the new derived class. This is similar to a relationship representing a child and a grandfather.

```python
# base class
class Grandfather:
    def __init__(self,grandfather_name):
        self.grandfather_name=grandfather_name

# intermediate class
class Father(Grandfather):
    def __init__(self,father_name,grandfather_name):
        super().__init__(grandfather_name)
        self.father_name=father_name

# derived class
class Son(Father):
    def __init__(self,son_name,father_name,grandfather_name):
        super().__init__(father_name,grandfather_name)
        self.son_name=son_name

child=Son('tommy','john','michael')
print(child.son_name) # son name
print(child.father_name) # father name
print(child.grandfather_name) # grandfather name


```

Explanation:
- Grandfather is the base class that stores the grandfather_name attribute.
- Father inherits from Grandfather and uses super().__init__(grandfather_name) to initialize the grandfather’s constructor before storing father_name.
- Son inherits from Father and again uses super().__init__(father_name, grandfather_name) to initialize both parent classes.
- When the Son object is created, constructors are called in the order: Son → Father → Grandfather.
- As a result, the child object contains all three attributes: son_name, father_name, and grandfather_name.

Hierarchical Inheritance
- When more than one derived class are created from a single base this type of inheritance is called hierarchical inheritance. In this program, we have a parent (base) class and two child (derived) classes.

```python
# base class 
class Father:
    def __init__(self,father_name):
        self.father_name=father_name

class child1(Father):
    def __init__(self,child_name,father_name):
        super().__init__(father_name)
        self.child_name=child_name

class child2(Father):
    def __init__(self,child_name,father_name):
        super().__init__(father_name)
        self.child_name=child_name

class child3(Father):
    def __init__(self,child_name,father_name):
        super().__init__(father_name)
        self.child_name=child_name

child1=child1('tommy','john')
child2=child2('jerry','john')

print(child1.child_name) # child1 name
print(child1.father_name) # father name

print(child2.child_name) # child2 name
print(child2.father_name) # father name
```

Explanation:
- Father is the base class that stores the father_name attribute for all children.
- child1, child2, and child3 inherit from Father and each uses super().__init__(father_name) to initialize the parent class constructor.
- Each child class also stores its own child_name attribute separately.
- When child1('tommy', 'john') and child2('jerry', 'john') are created, both objects receive the same father name but different child names.
- This is an example of hierarchical inheritance, where multiple child classes inherit from one parent class.

##### POLYMORPHISM

Polymorphism means "many forms".
It means the same method or function can work in different ways depending on the object using it.

Simple Nepal example:

Suppose an online shopping app in Nepal has different payment options like:

- eSewa
- Khalti
- Mobile Banking
- Debit/Credit Card

All of them use the same method called pay() but each payment type works differently.

- eSewa payment opens eSewa app or QR
- Khalti payment redirects to Khalti gateway
- Mobile Banking asks for bank verification
- Card payment checks card details and OTP

So, the method name is same (pay()), but behavior changes based on payment type.
This is called Polymorphism.

Types of Polymorphism
- Polymorphism refers to ability of the same method or operation to behave differently based on object or context. It mainly includes compile-time and runtime polymorphism.

1. Compile-time Polymorphism
- Compile-time polymorphism means deciding which method or operation to run during compilation, usually through method or operator overloading.

- Languages like Java or C++ support this. But Python doesn’t because it’s dynamically typed it resolves method calls at runtime, not during compilation. So, true method overloading isn’t supported, though similar behavior can be achieved using default or variable arguments.

2. Runtime polymorphism

- Runtime polymorphism means that the behavior of a method is decided while program is running, based on the object calling it. 
- This happens through Method Overriding a child class provides its own version of a method already defined in the parent class.

##### Polymorphism in Built-in Functions

Python’s built-in functions like len() and max() are polymorphic they work with different data types and return results based on type of object passed. This showcases it's dynamic nature, where same function name adapts its behavior depending on input

```python
print(len("hello guys"))
print(len([1,2,3]))
```

### Method overloading

Method overloading is a concept in object-oriented programming where a class can have multiple methods with the same name, but with different parameters or argument types. In Python, method overloading is not supported in the same way as it is in other object-oriented programming languages such as Java or C++. However, there are some ways to achieve similar functionality in Python.

1. Approach 1-use default arguments


```python
class calculator:
    
    def add(self,num1,num2=None,num3=None):
        if num2 is not None and num3 is not None:
            return num1 + num2 + num3
        elif num2 is not None:
            return num1 + num2
        else:
            return num1
        
calc=calculator()
print(calc.add(5))  # Output: 5
print(calc.add(5, 10))  # Output: 15
print(calc.add(5, 10, 15))  # Output: 30
```

**Explanation of the code:**
- The add() method uses default arguments (num2=None, num3=None) to simulate method overloading in Python.
- If all three numbers are provided, it returns the sum of three numbers, and if only two numbers are provided, it returns the sum of two numbers.
- If only one number is passed, it simply returns that number.
- This allows the same method add() to behave differently based on the number of arguments provided.

2. variable length argments


```python
class calculator:
    def add(self,*args):
        if len(args) == 1:
            return args[0]
        elif len(args) == 2:
            return args[0] + args[1]
        elif len(args) == 3:
            return args[0] + args[1] + args[2]
        else:
            return "Invalid number of arguments"
        
calc=calculator()
print(calc.add(5))  # Output: 5
print(calc.add(5, 10))  # Output: 15
print(calc.add(5, 10, 15))  # Output: 30
print(calc.add(5, 10, 15, 20))  # Output: Invalid number of arguments
```

**Explanation of the code:**
- This calculator class defines an add method that uses *args to accept a variable number of arguments.
- Inside the method, it checks how many arguments are passed using len(args).
- If there is one argument, it returns that value; if there are two, it returns their sum; and if there are three, it returns the sum of all three.
- If more than three arguments are given, it returns an error message saying “Invalid number of arguments.”
- This approach demonstrates how Python uses *args to achieve flexible method behavior similar to method overloading

### Method Overriding

Method overriding refers to the ability of a subclass to provide its own implementation of a method that is already defined in its parent class. When a method in a subclass has the same name and parameters as a method in its parent class, the subclass method will override the parent class method.

To override a method in Python, the subclass method must have the same name and parameters as the parent class method. When an object of the subclass calls the method, it will use the implementation provided in the subclass, rather than the one in the superclass.

```python
class Animal:
    def sound(self):
        print("Making some sound...")

class Dog(Animal):
    def sound(self, sound):
        print("Barking!", sound)

class Cat(Animal):
    def sound(self):
        print("Meowing...")

# Create some objects and call the sound method
animal = Animal()
dog = Dog()
cat = Cat()

animal.sound()  # Output: Making some sound...
dog.sound("Woof!")  # Output: Barking! Woof!
cat.sound()  # Output: Meowing...
```

**Explanation of the code:**
- In this example, we define a superclass Animal with a method sound that prints a generic sound. We then define two subclasses, Dog and Cat, that override the sound method with their own implementation

The super() function can be used to call the overridden method in the parent class, so that the parental behaviour is not lost.

```python
class Animal:
    def sound(self):
        print("Making some sound...")

class Dog(Animal):
    def sound(self, sound):
        print("Barking!", sound)
        super().sound()  # Call the parent class method

class Cat(Animal):
    def sound(self):
        print("Meowing...")
        super().sound()  # Call the parent class method

# Create some objects and call the sound method
animal = Animal()
dog = Dog()
cat = Cat()

animal.sound()  # Output: Making some sound...
dog.sound("Woof!")  # Output: Barking! Woof!
cat.sound()  # Output: Meowing...
```

Access Specifiers
- Access specifiers define how class members (variables and methods) can be accessed from outside the class. They help in implementing encapsulation by controlling the visibility of data. There are three types of access specifiers:


#### 1. Public Members
Public members are variables or methods that can be accessed from anywhere inside the class, outside the class or from other modules. By default, all members in Python are public. They are defined without any underscore prefix (e.g., self.name).

```python
class Employee:
    def __init__(self, name):
        self.name = name   # public attribute

    def display_name(self):   # public method
        print(self.name)

emp = Employee("John")
emp.display_name()   # Accessible
print(emp.name)      # Accessible
```

- **self.name:** Declared without underscores, so it is public.
- **display_name()**: Public method that prints the value of the public attribute.
- **emp.name:** Directly accessed from outside the class, showing public members are fully accessible.

##### 2. Protected members
Protected members are variables or methods that are intended to be accessed only within the class and its subclasses. They are not strictly private but should be treated as internal. In Python, protected members are defined with a single underscore prefix (e.g., self._name).

```python
class Employee:
    def __init__(self, name, age):
        self.name = name       # public
        self._age = age        # protected

class SubEmployee(Employee):
    def show_age(self):
        print("Age:", self._age)   # Accessible in subclass

emp = SubEmployee("Ross", 30)
print(emp.name)        # Public accessible
emp.show_age()         # Protected accessed through subclass
```

```python
print(emp._age)         # Accessible but not recommended (convention)
```

**Explanation:**

**self._age:** Defined with a single underscore, marking it as protected.
**SubEmployee:** Inherits from Employee and can access _age directly.
**Protected members:** should not be accessed outside the class hierarchy, but Python does not enforce this rule strictly.

##### 3. Private members

Private members are variables or methods that cannot be accessed directly from outside the class. They are used to restrict access and protect internal data. In Python, private members are defined with a double underscore prefix (e.g., self.__salary).

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self.__salary = salary    # private

    def show_salary(self):
        print("Salary:", self.__salary)

emp = Employee("Robert", 60000)
print(emp.name)          # Public accessible
emp.show_salary()        # Accessing private correctly
# print(emp.__salary)    # Error: Not accessible directly
```

**Explanation:**

- **self.__salary:** Defined with double underscores, so it is private.
- **show_salary():** A public method that provides safe access to the private attribute.
- **Attempting emp.__salary:** causes an AttributeError, proving private members cannot be accessed directly.

##### Getter and Setter Methods

In Python, getter and setter methods are used to access and modify private attributes safely. Instead of accessing private data directly, these methods provide controlled access, allowing you to:

- Read data using a getter method.
- Update data using a setter method with optional validation or restrictions.

```python
class Employee:
    def __init__(self):
        self.__salary = 50000  # Private attribute

    def get_salary(self):    # Getter method
        return self.__salary

    def set_salary(self, amount):   # Setter method
        if amount > 0:
            self.__salary = amount
        else:
            print("Invalid salary amount!")

emp = Employee()
print(emp.get_salary())  # Access salary using getter

emp.set_salary(60000)   # Update salary using setter
print(emp.get_salary())
```

**Explanation:**

- **__salary** is a private attribute, so it can't be accessed directly from outside the class.
- **get_salary()** is a getter method that safely returns the current salary.
- **set_salary(amount)** is a setter method that updates the salary only if the amount is positive.
- The object emp uses these methods to access and modify the salary while keeping the data protected.

#### Data Abstraction in Python

Data abstraction means showing only the essential features and hiding the complex internal details. In Python, abstraction is used to hide the implementation details from the user and expose only necessary parts, making the code simpler and easier to interact with.

##### Abstract Base Class
- In Python, an Abstract Base Class (ABC) is used to achieve data abstraction by defining a common interface for its subclasses. 
- It cannot be instantiated directly and serves as a blueprint for other classes.

Abstract classes are created using abc module and @abstractmethod decorator, allowing developers to enforce method implementation in subclasses while hiding complex internal logic.

```python
from abc import ABC, abstractmethod

class Greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass  # Abstract method

class English(Greet):
    def say_hello(self):
        print("Hello!")

g = English()
g.say_hello()
```

**Explanation:**

- Greet is an abstract class with a method say_hello() that has no implementation.
- English implements this method and returns a greeting.
- This keeps structure fixed while letting subclasses define their own behavior.

```python
class Bank_payment(ABC):
    @abstractmethod
    def pay(self):
        pass

class credit_card(Bank_payment):
    def pay(self):
        print("Payment made using credit card")

class mobile_banking(Bank_payment):
    def pay(self):
        print("Payment made using mobile banking")

cc = credit_card()
cc.pay()
mb = mobile_banking()
mb.pay()

```

##### Abstract Method

Abstract methods are method declarations without a body defined inside an abstract class. They act as placeholders that force subclasses to provide their own specific implementation, ensuring consistent structure across derived classes

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass  # Abstract method, no implementation here
```

**Explanation:** `make_sound()` is an abstract method in Animal class, so it doesn't have any code inside it.
