```python
# 1.label encoding----it should BE USED in target column only
# 2.one hot encoding---for large categorical variable,we donnot use one hot encoding because it increases the dimension of the data
# 3 Target encoding---for features columns

## todo task : implement encoding techniques on categorical columns
```

```python
df['Sex']=df['Sex'].map({'male':1,'female':0})
df.head()
```

```python
df['Embarked']=df['Embarked'].map({'S':0,'C':1,'Q':2})
df.head()
```

```python
X=df.drop('Survived',axis=1)
X.head()
```

```python
y=df['Survived']
y.head()
```

```python
# from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=40)
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
# from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(X_train,y_train)
```

```python
lr_pred=lr.predict(X_test)
lr_pred
```

```python
## for the performance metrics
#1.accuracy_score
#2.confusion_matrix
#3.classification_report
#4. precision
#5.recall
#6.f1_score
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,precision_score,recall_score,f1_score
```

```python
print("accuracy_score:",accuracy_score(y_test,lr_pred))
print("confusion_matrix:\n",confusion_matrix(y_test,lr_pred))
print("classification_report:\n",classification_report(y_test,lr_pred))
print("precision_score:",precision_score(y_test,lr_pred))
print("recall_score:",recall_score(y_test,lr_pred))
print("f1_score:",f1_score(y_test,lr_pred))
```

```python
### Accuracy--- out of your total predictions how many predictions are correct
### precision--- out of your total predicted positives how many are actual positives
# --- out of your total survived passanger how many are predicted correctly by your model

### recall---out of your total actual positives how many are predicted correctly by your model
```

```python
from sklearn.tree import DecisionTreeClassifier,ExtraTreeClassifier
```

```python
dtc=DecisionTreeClassifier(max_depth=5,criterion='gini')
dtc.fit(X_train,y_train)
dtc_pred=dtc.predict(X_test)
dtc_pred
```

```python
print("accuracy_score:",accuracy_score(y_test,dtc_pred))
print("confusion_matrix:\n",confusion_matrix(y_test,dtc_pred))
print("classification_report:\n",classification_report(y_test,dtc_pred))
print("precision_score:",precision_score(y_test,dtc_pred))
print("recall_score:",recall_score(y_test,dtc_pred))
print("f1_score:",f1_score(y_test,dtc_pred))
```

```python
etc=ExtraTreeClassifier(max_depth=5,criterion='gini')
etc.fit(X_train,y_train)
etc_pred=etc.predict(X_test)
etc_pred
```

```python
print("accuracy_score:",accuracy_score(y_test,etc_pred))
print("confusion_matrix:\n",confusion_matrix(y_test,etc_pred))
print("classification_report:\n",classification_report(y_test,etc_pred))
print("precision_score:",precision_score(y_test,etc_pred))
print("recall_score:",recall_score(y_test,etc_pred))
print("f1_score:",f1_score(y_test,etc_pred))
```
